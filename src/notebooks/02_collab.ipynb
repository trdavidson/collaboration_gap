# Collaborative games

Two models each see part of the maze and must agree on a route through dialogue.
Self-play pairs give the collaborative bars of the paper's Figure 4.1; mixed
pairs give the heatmaps of Figure 4.3.

```bash
python -m src.create_maze_jobs --config_file experiments/job_configs/example_collab.yaml
bash experiments/jobs/example_collab/run_all_jobs.sh
```

````python
from notebook_helpers import set_working_dir_to_root
set_working_dir_to_root()

import pandas as pd
from src.utils.experiment_loading import get_experiment_data, add_derived_columns
from src.utils.experiment_loading import load_solution_visual
from src.utils.plotting_utils import plot_heatmap
````

````python
df = get_experiment_data(['example_collab'], add_messages=False, return_as_df=True)
add_derived_columns(df)
print(f'{len(df)} games')
````

````python
df.groupby(['lm_pair_pp', 'split_1_proportion'])[
    ['weighted_outcome', 'outcome_success']].agg(['mean', 'sem', 'size'])
````

With several models in `lm_pair`, the heatmap puts agent 1 on one axis and
agent 2 on the other.

````python
plot_heatmap(
    df.copy(),
    col_1='lm_1_name_pp',
    col_2='lm_2_name_pp',
    value_col='weighted_outcome',
    panel_col='split_1_proportion',
    title_col='Weighted outcome',
    show_ci=True,
);
````

One game: the maze with the agreed route drawn on it. Pass
