# Solo baseline (one-shot)

A single model gets the maze and proposes a route in one answer, optionally
followed by a critic step that reviews it. Two conditions, as in the paper's
solo baseline: the full maze, and the *distributed* maze (`split_view=True`),
where the same information arrives as the two partial maps the collaborating
agents receive.

```bash
python -m src.create_maze_jobs --config_file experiments/job_configs/example_one_shot.yaml
bash experiments/jobs/example_one_shot/run_all_jobs.sh
python -m src.create_maze_jobs --config_file experiments/job_configs/example_one_shot_critic.yaml
bash experiments/jobs/example_one_shot_critic/run_all_jobs.sh
```
